*EXPLORATORY DATA ANALYSIS*

GENDER BASED

```python
df.columns
```

```python
ax= sns.countplot(x = 'Gender', data=df, palette='viridis', hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

```python
sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Gender', y='Amount', data=sales_gen,  palette='viridis', hue='Gender')
plt.show()
```

From the graphs above we can conclude that the most of the buyers are females and the purchasing powers of females are higher than that of male.

AGE BASED

```python
age_order = sorted(df['Age Group'].unique())
print(age_order)
```

```python
sorted_order= ['0-17', '18-25', '26-35', '36-45', '46-50', '51-55', '55+']
ax = sns.countplot(x= 'Age Group', data= df,  palette='Set1', hue='Gender', order= sorted_order)
for bars in ax.containers:
  ax.bar_label(bars)
plt.show()
```

As we can see that the most of the buyers are in the age group of 26-35. And most of them are females.

```python
sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Age Group', y='Amount', data=sales_age,  palette='Set2', hue='Age Group')
plt.show()
```

People aged between 26-35 have more purchasing power.

```python
df.columns
```

STATE BASED

```python
sales_state= df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by= 'Orders', ascending=False).head(10)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='State', y='Orders', data=sales_state,  palette='Set2', hue='State')
plt.show()
```

most of the orders are from Uttar Pradesh, Maharastra, and Karnataka.

```python
sales_state= df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by= 'Amount', ascending=False).head()
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='State', y='Amount', data=sales_state,  palette='Set2', hue='State')
plt.show()
```

```python
sales_marr= sns.countplot(x= 'Marital_Status', data=df, palette='Set2', hue='Marital_Status')
sns.set(rc={'figure.figsize':(7,5)})
for bars in sales_marr.containers:
    sales_marr.bar_label(bars)
plt.show()
```

```python
df.columns
```

```python
sales_marr= df.groupby(['Marital_Status','Gender'], as_index=False)['Amount'].sum().sort_values(by= 'Amount', ascending=False).head()
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='Marital_Status', y='Amount', data=sales_marr,  palette='Set2', hue='Gender')

plt.show()
```

here we can observe that the married people ourchased more than the unmarried ones in which females are more.

`Occupation based

```python
ax= sns.countplot(data= df, x='Occupation', palette='viridis', hue='Occupation')
for bars in ax.containers:
    ax.bar_label(bars)

plt.show()
```

```python
sales_occ= df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by= 'Amount', ascending=False).head()
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(x='Occupation', y='Amount', data=sales_occ,  palette='viridis', hue='Occupation')

plt.show()
```

as we can see from the graph that the more consumers are from IT sector with maximum purchasing power. 

Product category


```python
sns.set(rc={'figure.figsize':(20,5)})
top_10= df['Product_Category'].value_counts().head(10).index
ax= sns.countplot(data= df, x='Product_Category', palette='viridis', hue='Product_Category', order=top_10 )
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

from the graph above colthing, food, and electronics items were were sold more in quantity.

```python
sales_catg= df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by= 'Amount', ascending=False).head()
sns.set(rc={'figure.figsize':(10,5)})
sns.barplot(x='Product_Category', y='Amount', data=sales_catg,  palette='viridis', hue='Product_Category')

plt.show()
```

but here we can see more amount were spent on food than clothing even though cloths consumers were more.

*Conclusion:
Married women age group 26-35 yrs from UP, Maharastra and Karnataka working in IT, Healthcare and Aviation are more likely to buy products from Food, Clothing and Electronics category.*
